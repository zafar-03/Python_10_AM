```python
import matplotlib.pyplot as plt
```

## Line Chart:

```python
months = ["Jan","Feb","Mar","Apr","May"]
sales = [1000,1500,700,2000,2200]
plt.plot(months,sales,marker="o")

plt.title("Company Sales")
plt.xlabel("Month")
plt.ylabel("No. of Products")
plt.show()
```

## Bar Chart :

```python
values = [1,2,3,4,5]
marks = [45,67,89,45,23]

plt.bar(values,marks)
# plt.bar(marks,values)
```
