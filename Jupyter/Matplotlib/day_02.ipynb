## Stack Plot:

```python
import matplotlib.pyplot as plt 
```

```python
no_of_days = [1,2,3,4,5,6,7]

maths  = [2,4,3,5,2,4,3]
python  = [1,4,2,4,6,3,2]
c = [7,4,5,2,1,3,2]


plt.stackplot(no_of_days,maths,c,python)

plt.show()
```

```python
no_of_days = [1,2,3,4,5,6,7]

maths   = [2,4,3,5,2,4,3]
python  = [1,4,2,4,6,3,2]
c       = [7,4,2,1,1,3,2]


plt.stackplot(no_of_days,maths,c,python,labels=["maths","c","python"])

plt.legend()
plt.xlabel("Day")
plt.ylabel("no. of Students")
plt.title("Student Analysis")

plt.savefig("stackplot.png")
plt.show()
```
