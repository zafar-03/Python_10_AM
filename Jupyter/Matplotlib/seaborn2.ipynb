```python
import matplotlib.pyplot as plt 
import seaborn as sbn
```

```python
my_data = sbn.load_dataset("penguins")

sbn.lineplot(data=my_data,x="species",y="body_mass_g")
# sbn.set_style("ticks")

plt.show()
```

```python
sbn.lineplot(data=my_data,x="species",y="body_mass_g")
sbn.set_style("dark")

plt.show()
```

```python
sbn.lineplot(data=my_data,x="species",y="body_mass_g")
sbn.set_style("ticks")

plt.show()
```

```python
sbn.lineplot(data=my_data,x="species",y="body_mass_g")
sbn.set_style("whitegrid")

plt.show()
```

```python
sbn.lineplot(data=my_data,x="species",y="body_mass_g")
sbn.despine()

plt.show()
```

```python
sbn.lineplot(data=my_data,x="species",y="body_mass_g")
sbn.set_style("darkgrid")
sbn.despine()

plt.show()
```

```python
sbn.lineplot(data=my_data,x="species",y="body_mass_g")
sbn.despine(left=False)

plt.show()
```

```python
sbn.lineplot(data=my_data,x="species",y="body_mass_g")
sbn.despine(left=True)

plt.show()
```

```python
sbn.scatterplot(data=my_data,x="species",y="body_mass_g")

plt.show()
```

```python
sbn.scatterplot(data=my_data,x="body_mass_g",y="species")

plt.show()
```

```python
sbn.scatterplot(data=my_data,x="bill_length_mm",y="body_mass_g",hue="island")

plt.show()
```

```python
sbn.set_style("darkgrid")
sbn.scatterplot(data=my_data,x="bill_length_mm",y="body_mass_g",hue="island")

plt.show()
```

```python
sbn.set_style("darkgrid")
sbn.scatterplot(data=my_data,x="bill_length_mm",y="body_mass_g",hue="island")
sbn.despine(top=True, right=True, left=False, bottom=False, offset=None, trim=False, ax=None)

plt.show()
```

```python
sbn.set_style("darkgrid")
sbn.scatterplot(data=my_data,x="bill_length_mm",y="body_mass_g",hue="island")
sbn.despine(top=False, right=False, left=True, bottom=True, offset=None, trim=False, ax=None)

plt.show()
```
