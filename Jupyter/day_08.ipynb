## Marge : 

```python
df1 = pd.DataFrame(
    {
        "Id" : [1,2,3],
        "Name" : ["Raj","Sahil","Rajesh"]
    })

df2 = pd.DataFrame(
    {
        "Id" : [1,2,3],
        "Marks" : [67,89,45],
        "Age" : [12,23,16]
    })
```

```python
df1 
```

```python
df2 
```

```python
newdf = pd.merge(df1,df2)
newdf
```

```python
newdf = pd.merge(df1,df2,how="inner")
newdf
```

```python
newdf = pd.merge(df1,df2,how="outer")
newdf
```

```python
newdf = pd.merge(df1,df2,how="left")
newdf
```

```python

df1.join(df2,on="Id")
```
