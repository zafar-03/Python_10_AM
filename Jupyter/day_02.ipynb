# Slicing :

```python
a3 = np.array([11,2,3,4,3,7,88,4,11,33])
print(a3)
```

```python
print(a3[1:])
```

```python
print(a3[1:4])
```

```python
print(a3[5:])
```

```python
print(a3[5:8])
```

```python
print(a3[1:8])
```

```python
print(a3[1:8:1])
```

```python
print(a3[1:8:2])
```

```python
print(a3[1:8:5])
```

```python
print(a3[1:8:-1])
```

```python
print(a3[1:8:-3])
```

```python
print(a3[-1:])
```

```python
print(a3[-1:-5])
```

```python
print(a3[-5:-1])
```
