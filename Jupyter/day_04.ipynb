```python
a1 =  np.array([6,2,3])
a2 = np.array([5,4,3])
print(a1)
print(a2)
```

```python
print(a1 == a2)
```

```python
print(a1 != a2)
```

```python
print(a1 < a2)
```

```python
print (a1 <= a2)
print (a1 > a2)
print (a1 >= a2)
```

```python
print(np.logical_and(a1 > a2,a1!=a2))
# logical_or
```
