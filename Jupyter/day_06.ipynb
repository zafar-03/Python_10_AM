# Creation of DataFrames:


```python
# 1. List of List 
my_list = [
    ["Raj",10,2],
    ["Sahil",20,4],
    ["Rahul",30,5]
]
df1 = pd.DataFrame(my_list,columns=["name","age","std"])
print(df1)
```

```python
# 2. Dict 
my_dict = {
    "name" : ["Raj","Sahil","Rahul"],
    "age" : [10,23,43],
    "std" : [2,3,4]
}
df2 = pd.DataFrame(my_dict)
print(df2)
```

```python
# 3. List of Dict
my_list = [
    {"name" : "Raj","age": 10, "std" :2},
    {"name" : "Sahil","age": 10, "std" :2},
    {"name" : "Rahul","age": 10, "std" :2}
]
df3 = pd.DataFrame(my_list,index=["@","!","%"])
print(df3)
```

```python
print(df3.loc["@"])
```

```python
print(df3.iloc[0])
```
