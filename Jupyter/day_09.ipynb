##  merge : 

```python
import pandas as pd
```

```python
df1 = pd.DataFrame(
    {
        "Id" : [1,2,3,5],
        "Name" : ["Raj","Sahil","Rajesh","Rajveer"]
    })

df2 = pd.DataFrame(
    {
        "Id" : [1,2,3,4],
        "Marks" : [67,89,45,60],
        "Age" : [12,23,16,20]
    })
```

```python
df1
```

```python
df2
```

```python
pd.merge(df1,df2,on="Id")
```

```python
pd.merge(df1,df2,on="Id",how="inner")
```

```python
pd.merge(df1,df2,on="Id",how="outer")
```

```python
pd.merge(df1,df2,on="Id",how="left")
```

```python
pd.merge(df1,df2,on="Id",how="right")
```

```python
df = pd.DataFrame(
    {
        "Name" : ["Raj","Sahil","Rajesh","Rajveer"],
        "Marks" : [67,89,45,60],
        "Age" : [12,23,16,20]
    })
```

```python
df
```

```python
df.pivot(index="Name",columns="Marks",values="Age")
```

```python
df =  pd.DataFrame({
    "Date": ["2026-01-01", "2026-01-01", "2026-01-01", "2026-01-02"],
    "City": ["Chicago", "Austin", "Chicago", "Austin"],
    "Temperature": [32, 75, 30, 78]
})
```

```python
df.pivot(index="City",columns="Date",values="Temperature")
```

```python
output =  df.pivot_table(index="Date",columns="City",values="Temperature",aggfunc="sum")
```
