# Rename Specific Column name

```python
import pandas as pd
```

```python
my_list = [
    {"name" : "Raj","age": 10, "std" :2},
    {"name" : "Sahil","age": 10, "std" :2},
    {"name" : "Rahul","age": 10, "std" :2}
]
df = pd.DataFrame(my_list)
print(df)
```

```python
df.rename(columns={"std" : "standard"},inplace=True)
```

```python
print(df)
```

## Read and Write :

### 1. CSV File:

### Write : 

```python
df.to_csv("csvfile.csv",index=False)
```

```python
df.to_csv("csvfile2.csv",index=True)
```

## Handle missing values: 

```python
my_list = [
    {"name" : "Raj","age": 23, "std" :2},
    {"name" : "Sahil","age": 10, "std" :2},
    {"name" : "Rahul","age": 10, "std" :4}
]
df = pd.DataFrame(my_list)
print(df)
```

```python
df.isnull()
```

```python
print(df.isnull().sum())
```

```python
df.dropna()
```

```python
df
```

### duplicates : 

```python
print(df.duplicated("age"))
```

```python
df.drop_duplicates("std")
```

```python
df.dtypes
```

```python
df["age"] = df["age"].astype("str")
```
