```python
df1 = pd.DataFrame({
    "id" : [1,2,3],
    "name" : ["Raj","Rahul","Ramesh"]
})

df2 = pd.DataFrame({
    "marks" : [67,90,56],
    "age" : [12,14,13]
},index=[1,2,"C"])
```

```python
df1
```

```python
df2
```

```python
df1.join(df2)
```

```python
#   inner,outer ,left(default),right

# join : index (always)
# merge: (identity)
# pd.merge(df1,df2,on="",how=)

df1.join(df2,how="left")
```
